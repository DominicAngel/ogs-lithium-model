```python
from pathlib import Path
import subprocess

input_dir = Path(r"E:\ADATA\LITHIUM\OGS\OGS_files\input")
ogs_bin = Path(r"C:\Users\dominic.becerra\Documents\OGS\ogs\bin")

remove_unused = ogs_bin / "RemoveUnusedPoints.exe"
identify = ogs_bin / "identifySubdomains.exe"

for fname in ["left.vtu", "right.vtu"]:
    in_file = input_dir / fname
    clean_file = input_dir / f"clean_{fname}"

    result = subprocess.run(
        [
            str(remove_unused),
            "-i", str(in_file),
            "-o", str(clean_file)
        ],
        cwd=input_dir,
        capture_output=True,
        text=True
    )

    print(fname, "RemoveUnusedPoints")
    print(result.stdout)
    print(result.stderr)

    result = subprocess.run(
        [
            str(identify),
            "-m", str(input_dir / "mesh5.vtu"),
            "-f",
            str(clean_file)
        ],
        cwd=input_dir,
        capture_output=True,
        text=True
    )

    print(fname, "identifySubdomains")
    print(result.stdout)
    print(result.stderr)
```
