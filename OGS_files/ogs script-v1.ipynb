```python
# ==================================================
# CLEAN OGS MINIMAL LIQUID FLOW MODEL
# Dom's lithium project, first OGS test
# ==================================================

import os
import shutil
import subprocess
from pathlib import Path
```

```python
# --------------------------------------------------
# 1. Define folders
# --------------------------------------------------

base_dir = Path(r"E:\ADATA\LITHIUM\OGS\OGS_files")
input_dir = base_dir / "input"
output_dir = base_dir / "output"

input_dir.mkdir(parents=True, exist_ok=True)
output_dir.mkdir(parents=True, exist_ok=True)

print("Input folder:", input_dir)
print("Output folder:", output_dir)
```

```python
# --------------------------------------------------
# 2. Copy mesh into OGS input folder
# --------------------------------------------------

source_mesh = Path(r"Jupyter notebooks/mesh6.vtu")
target_mesh = input_dir / "mesh6.vtu"

shutil.copy(source_mesh, target_mesh)

print("Mesh copied to:", target_mesh)
```

```python
# --------------------------------------------------
# 4. Calling OGS project file
# --------------------------------------------------

input_dir = Path(r"E:\ADATA\LITHIUM\OGS\OGS_files\input")

prj_path = input_dir / "simple_flow.prj"
mesh_path = input_dir / "mesh6.vtu"

print("PRJ exists:", prj_path.exists())
print("Mesh exists:", mesh_path.exists())
print("PRJ path:", prj_path)
```

```python
# --------------------------------------------------
# 5. Run OGS
# --------------------------------------------------
ogs_exe = Path(r"C:\Users\dominic.becerra\Documents\OGS\ogs\bin\ogs.exe")


result = subprocess.run(
    [str(ogs_exe), str(prj_path)],
    cwd=output_dir,
    capture_output=True,
    text=True
)

print(result.stdout)
print(result.stderr)
```

```python
# --------------------------------------------------
# 6. Check output files
# --------------------------------------------------

print("Output files:")
print(os.listdir(output_dir))
```
