```python
import meshio
from pathlib import Path

input_dir = Path(r"E:\ADATA\LITHIUM\OGS\OGS_files\input")

for fname in ["left.vtu", "right.vtu"]:
    m = meshio.read(input_dir / fname)
    print("\n", fname)
    print("Number of points:", len(m.points))
    print("Cells:", m.cells)
    print("Point data:", m.point_data.keys())
    print("Cell data:", m.cell_data_dict.keys())
```

```python
import numpy as np
import meshio
from pathlib import Path

input_dir = Path(r"E:\ADATA\LITHIUM\OGS\OGS_files\input")

for fname in ["left.vtu", "right.vtu"]:
    m = meshio.read(input_dir / fname)
    tri = m.cells_dict["triangle"]
    used_points = np.unique(tri.flatten())

    print("\n", fname)
    print("Total stored points:", len(m.points))
    print("Used points by triangles:", len(used_points))
    print("First 10 used point ids:", used_points[:10])
```
