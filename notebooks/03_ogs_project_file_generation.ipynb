# OGS Project File Generation: Simple Model

    This notebook creates the main OpenGeoSys (OGS) project files required to define and run a simplified subsurface flow simulation.

    OGS requires the path to a .prj-file as an argument; this is the central configuration file of the numerical model. It specifies how the simulation is assembled by linking together the model geometry, computational meshes, physical processes, material properties, boundary conditions, numerical solvers, time stepping strategy, and output settings.

    The notebook is organized into the following main sections:

        1. Geometry file generation
        2. OGS project file construction
        3. File export and validation
        4. Project file validation

## 1. Geometry file generation

    OGS uses geometry files (`.gml`) to define geometrical entities that can later be referenced by boundary conditions, source terms, or additional model definitions.

    A `.gml` file contains → points, lines, surfaces

    For this simplified example, a minimal geometry file is generated programmatically.

```python
from pathlib import Path

input_dir = Path("../input")
```

```python
# ---------------------------------------------------------------
# Define a minimal OpenGeoSys geometry structure (.gml file).
#
# The .gml file normally stores geometric reference entities (points, polylines, surfaces) that can be used by OGS for
# boundary conditions, source terms, or observation locations.

# In this workflow, material assignments and boundary conditions are controlled through Gmsh physical groups and extracted .vtu
# submeshes rather than geometry entities.
#
# Therefore, the .gml file is included mainly for completeness and compatibility within the example OGS workflow. 
# ---------------------------------------------------------------

gml_content = """<?xml version="1.0" encoding="ISO-8859-1"?>
<OpenGeoSysGLI>
    <name>simple_geometry</name>
    <points>
    </points>
    <polylines>
    </polylines>
</OpenGeoSysGLI>
"""

gml_path = input_dir / "simple_geometry.gml"
gml_path.write_text(gml_content)

print("Geometry file created:", gml_path)
```

### 4. Project file validation

    Before running the simulation, the generated `.prj` file is validated against the official OpenGeoSys XML schema (`OpenGeoSysProject.xsd`).

    This step verifies that the project structure is syntactically correct and consistent with the format expected by OGS, helping detect XML construction errors during notebook development.

```python
from lxml import etree
from pathlib import Path

prj_path = Path(r"E:\ADATA\LITHIUM\OGS\OGS_files\input\simple_flow.prj")
xsd_path = Path(r"C:\Users\dominic.becerra\Documents\OGS\ogs\bin\OpenGeoSysProject.xsd")

schema = etree.XMLSchema(etree.parse(str(xsd_path)))
doc = etree.parse(str(prj_path))

print("Valid XML against XSD:", schema.validate(doc))

for error in schema.error_log:
    print(error.message)
```
