```python
import subprocess
import os               #Python interacts with the computer's file system.
```

*Subprocess - MESH GENERATION*

```python

mesh_folder = r"C:\Users\dominic.becerra\Documents\OGS\MESH"

cmd = [
    r"C:\Users\dominic.becerra\Documents\OGS\OGS\bin\generateStructuredMesh.exe",
    "-o", "layer.vtu",
    "-e", "hex",
    "--lx", "10",
    "--ly", "10",
    "--lz", "5",
    "--dz0", "1",
    "--mz", "0.8"
]

result = subprocess.run(cmd, capture_output=True, text=True, cwd=mesh_folder)       #cwd = run the command inside the folder "mesh_folder"
                                                                                    #"capture_output" = capture what the program prints
                                                                                    #"text" = store it as readable text

print(result.stdout)            #Shows the normal output produced by OGS
print(result.stderr)            #Shows any error messages produced by OGS.
print(result.returncode)        #success or failure code
```

```python
mesh_file = os.path.join(mesh_folder, "layer.vtu")

if os.path.exists(mesh_file):
    print("Mesh successfully created:")
    print(mesh_file)
else:
    print("Mesh file not found.")
```
