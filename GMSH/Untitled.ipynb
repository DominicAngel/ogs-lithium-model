```python
import csv
import gmsh
gmsh.initialize()
gmsh.model.add("mesh5.msh")

# ... your geometry and mesh creation here ...

# Export volumes with COM
volumes_info = []
for dim, tag in gmsh.model.getEntities(3):
    x, y, z = gmsh.model.occ.getCenterOfMass(dim, tag)
    volumes_info.append([tag, x, y, z])

# Write to CSV
with open("volumes_info.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["Volume Tag", "COM X", "COM Y", "COM Z"])
    writer.writerows(volumes_info)

gmsh.finalize()
```
